```python
#W8A3
try:
    n = int(input())
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)
        n = sum(int(c) ** 2 for c in str(n))
    print("YES" if n == 1 else "NO")
except ValueError:
    print("NO")

```

```python
#W8A4
a, b = input().split()
try:
    x = int(a)
    y = int(b)
except ValueError:
    print("Nhập sai kiểu dữ liệu")
else:
    print(x + y)
finally:
    print("Kết thúc chương trình.")
```
