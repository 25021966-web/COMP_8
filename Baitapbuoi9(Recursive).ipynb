```python
#W9A1
def merge(a, b):
    i = j = 0
    res = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            res.append(a[i]); i += 1
        else:
            res.append(b[j]); j += 1
    res.extend(a[i:])
    res.extend(b[j:])
    return res
```

```python
#W9A2
import heapq

def kth_largest(mat, k):
    h = []
    for row in mat:
        for x in row:
            if len(h) < k:
                heapq.heappush(h, x)
            elif x > h[0]:
                heapq.heapreplace(h, x)
    return h[0]
```

```python
#W9A3
def min_swap(num):
    s = list(num)
    for i in range(len(s)):
        m = min(s[i+1:], default=s[i])
        if m < s[i]:
            j = len(s) - 1 - s[::-1].index(m)
            s[i], s[j] = s[j], s[i]
            break
    return ''.join(s)
```

```python
#W9A4
def can_convert(a, b):
    n, m = len(a), len(b)
    dp = [[False]*(m+1) for _ in range(n+1)]
    dp[0][0] = True

    for i in range(1, n+1):
        if a[i-1].islower():
            dp[i][0] = dp[i-1][0]

    for i in range(1, n+1):
        for j in range(1, m+1):
            if a[i-1].upper() == b[j-1]:
                dp[i][j] = dp[i-1][j-1]
            if a[i-1].islower():
                dp[i][j] |= dp[i-1][j]
    return "YES" if dp[n][m] else "NO"
```

```python
#W9A5
def is_subarray(a, b):
    for i in range(len(b) - len(a) + 1):
        if b[i:i+len(a)] == a:
            return "YES"
    return "NO"
```

```python
#W9A6
def next_permutation(a):
    i = len(a) - 2
    while i >= 0 and a[i] >= a[i+1]:
        i -= 1
    if i < 0:
        return sorted(a)

    j = len(a) - 1
    while a[j] <= a[i]:
        j -= 1
    a[i], a[j] = a[j], a[i]
    a[i+1:] = reversed(a[i+1:])
    return a
```

```python
#W9A7
def canPlaceFlowers(flowerbed, k):
    for i in range(len(flowerbed)):
        if flowerbed[i] == 0:
            left = (i == 0 or flowerbed[i-1] == 0)
            right = (i == len(flowerbed)-1 or flowerbed[i+1] == 0)
            if left and right:
                flowerbed[i] = 1
                k -= 1
```

```python
#W9A8
def is_path_crossing(moves):
    x = y = 0
    visited = {(0, 0)}
    for m in moves:
        if m == 'R': x += 1
        if m == 'L': x -= 1
        if m == 'U': y += 1
        if m == 'D': y -= 1
        if (x, y) in visited:
            return True
        visited.add((x, y))
    return False
```

```python
#W9A9
def Median(nums):
    a, b, c, d, e = nums
    if a > b: a, b = b, a
    if c > d: c, d = d, c
    if a > c: a, c = c, a
    if b > d: b, d = d, b
    if b > c: b, c = c, b
    return max(b, min(c, e))
```

```python
#W9A10
def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()
    i = 0
    res = 0
    for h in houses:
        while i + 1 < len(heaters) and abs(heaters[i+1] - h) <= abs(heaters[i] - h):
            i += 1
        res = max(res, abs(heaters[i] - h))
    return res
```
