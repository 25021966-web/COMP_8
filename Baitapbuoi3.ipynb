```python
#1
n = int(input("Nhập số nguyên n: "))
print("Số đảo ngược:", str(n)[::-1])

#2
a = int(input("Nhập a: "))
b = int(input("Nhập b: "))
a = a ^ b
b = a ^ b
a = a ^ b
print("Sau khi hoán đổi: a =", a, ", b =", b)

#3
n = int(input("Nhập số n: "))
if n > 0 and (n & (n - 1)) == 0:
    print("True")
else:
    print("False")

#4
m = int(input("Nhập m: "))
n = int(input("Nhập n: "))
print("Kết quả m chia n làm tròn xuống:", m // n)

#5
import math
m = int(input("Nhập m: "))
n = int(input("Nhập n: "))
print("Kết quả m chia n làm tròn lên:", math.ceil(m / n))

#6
x = int(input("Nhập số nguyên dương x: "))
if x % 2 == 0:
    print("Even")
else:
    print("Odd")

#7
a = int(input("Nhập a: "))
b = int(input("Nhập b: "))
if a < 0 and b < 0:
    print("Yes")
else:
    print("No")

#8
a = input("Nhập chuỗi a: ")
b = input("Nhập chuỗi b: ")
print("True" if len(a) > len(b) else "False")

#9
a = int(input("Nhập cạnh a: "))
b = int(input("Nhập cạnh b: "))
c = int(input("Nhập cạnh c: "))
if a + b > c and a + c > b and b + c > a and a > 0 and b > 0 and c > 0:
    print("Yes")
else:
    print("No")

#10
a = int(input("Nhập số a: "))
b = int(input("Nhập số b: "))
c = int(input("Nhập số c: "))
d = int(input("Nhập số d: "))
print("Số lớn nhất là: ", max(a, b, c, d))

#11
a = int(input("Nhập cạnh a: "))
b = int(input("Nhập cạnh b: "))
c = int(input("Nhập cạnh c: "))

if a <= 0 or b <= 0 or c <= 0 or a + b <= c or a + c <= b or b + c <= a:
    print("Không phải tam giác")
else:
    if a == b == c:
        print("Tam giác đều")
    elif a == b or a == c or b == c:
        print("Tam giác cân")
    else:
        print("Tam giác thường")

#12
year = int(input("Nhập năm: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Yes")
else:
    print("No")

#13
kwh = int(input("Nhập số kWh điện tiêu thụ: "))
if kwh < 0:
    print("Số điện không hợp lệ!")
elif kwh <= 50:
    total = kwh * 1500
elif kwh <= 100:
    total = 50 * 1500 + (kwh - 50) * 2000
else:
    total = 50 * 1500 + 50 * 2000 + (kwh - 100) * 3000
print("Tiền điện phải trả:", total)

#14
a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
if a == 0:
    if b == 0:
        print("Vô số nghiệm")
    else:
        print("Vô nghiệm")
else:
    x = -b / a
    print("Nghiệm của phương trình là:", round(x, 2))

#15
score = float(input("Nhập điểm trung bình: "))
if score >= 8.0:
    print("Giỏi")
elif score >= 6.5:
    print("Khá")
elif score >= 5.0:
    print("Trung bình")
else:
    print("Yếu")

#16
x = float(input("Nhập số thực: "))
if x >= 0:
    down = int(x)
else:
    down = int(x) - (x != int(x))
if x >= 0:
    up = int(x) + (x != int(x))
else:
    up = int(x)
frac = x - int(x)
if x >= 0:
    nearest = int(x) + (frac >= 0.5)
else:
    nearest = int(x) - (frac <= -0.5)

print("Làm tròn xuống:", down, "| Làm tròn lên:", up, "| Gần nhất:", nearest)
```
