```python
#W5A1
def Max_Of_Two(a, b):
    return a if a > b else b
```

```python
#W5A2
def swap(a, b):
    return b, a
```

```python
#W5A3
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
#W5A4
def is_perfect(n):
    if n <= 1:
        return False
    s = 0
    for i in range(1, n):
        if n % i == 0:
            s += i
    return s == n
```

```python
#W5A5
def find_first_position(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1
```

```python
#W5A6
def factorial(n):
    res = 1
    for i in range(1, n + 1):
        res *= i
    return res
```

```python
#W5A7
def calculator(num1, num2, operat):
    if operat == '+':
        result = num1 + num2
    elif operat == '-':
        result = num1 - num2
    elif operat == '*':
        result = num1 * num2
    elif operat == '/':
        result = num1 / num2
    return round(result, 2)
```

```python
#W5A8
def hamming_distance(x, y):
    bx = bin(x)[2:]
    by = bin(y)[2:]

    max_len = max(len(bx), len(by))
    bx = bx.zfill(max_len)
    by = by.zfill(max_len)

    cnt = 0
    for i in range(max_len):
        if bx[i] != by[i]:
            cnt += 1
    return cnt
```

```python
#W5A9
def sum_of_digits(n):
    s = 0
    for c in str(n):
        s += int(c)
    return s
```

```python
#W5A10
def is_isomorphic(a, b):
    map_ab = {}
    map_ba = {}

    for i in range(len(a)):
        x = a[i]
        y = b[i]

        if x in map_ab and map_ab[x] != y:
            return False
        if y in map_ba and map_ba[y] != x:
            return False

        map_ab[x] = y
        map_ba[y] = x

    return True
```
