```python
# W8A1
class Cylinder:
    def __init__(self, r, h):
        self.r = r
        self.h = h

    def surface_area(self):
        return 2 * 3.14 * self.r * (self.r + self.h)

    def volume(self):
        return 3.14 * self.r * self.r * self.h


r, h = map(float, input().split())
c = Cylinder(r, h)
print(f"{c.surface_area():.2f} {c.volume():.2f}")

```

```python
# W8A2
class Date:
    def __init__(self, d, m, y):
        self.d = d
        self.m = m
        self.y = y

    def is_leap(self):
        return self.y % 400 == 0 or (self.y % 4 == 0 and self.y % 100 != 0)

    def is_valid(self):
        days = [31, 29 if self.is_leap() else 28, 31, 30, 31, 30,
                31, 31, 30, 31, 30, 31]
        return 1 <= self.m <= 12 and 1 <= self.d <= days[self.m - 1]

    def next_day(self):
        days = [31, 29 if self.is_leap() else 28, 31, 30, 31, 30,
                31, 31, 30, 31, 30, 31]
        self.d += 1
        if self.d > days[self.m - 1]:
            self.d = 1
            self.m += 1
            if self.m > 12:
                self.m = 1
                self.y += 1


d, m, y = map(int, input().split("/"))
date = Date(d, m, y)

if not date.is_valid():
    print("INVALID")
else:
    date.next_day()
    print(f"{date.d:02d}/{date.m:02d}/{date.y}")
```

```python
# W8A3 – Point
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def is_origin(self):
        return self.x == 0 and self.y == 0

    def on_x_axis(self):
        return self.y == 0

    def on_y_axis(self):
        return self.x == 0

    def distance_to_origin(self):
        return math.sqrt(self.x**2 + self.y**2)


x, y = map(float, input().split())
p = Point(x, y)

print("Yes" if p.is_origin() else "No")
print("Yes" if p.on_y_axis() else "No")
print("Yes" if p.on_x_axis() else "No")
print(f"{p.distance_to_origin():.2f}")
```

```python
# W8A4
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def set_numbers(self, a, b):
        self.a = a
        self.b = b

    def add(self): return self.a + self.b
    def subtract(self): return self.a - self.b
    def product(self): return self.a * self.b
    def divide(self): return self.a / self.b if self.b != 0 else "Error"
    def power(self): return self.a ** self.b
    def mod(self): return self.a % self.b


a, b = map(float, input().split())
calc = Calculator(a, b)

while True:
    op = input()
    if op == "add":
        print(calc.add())
    elif op == "subtract":
        print(calc.subtract())
    elif op == "product":
        print(calc.product())
    elif op == "divide":
        print(calc.divide())
    elif op == "power":
        print(calc.power())
    elif op == "mod":
        print(calc.mod())
    elif op == "set":
        a, b = map(float, input().split())
        calc.set_numbers(a, b)

    if input("Bạn có muốn thoát? ").lower() == "yes":
        break
```

```python
# W8A5
class ShoppingCart:
    def __init__(self):
        self.cart = []

    def add(self, name, price):
        self.cart.append({"name": name, "price": price})

    def remove(self, name):
        self.cart = [p for p in self.cart if p["name"] != name]

    def is_empty(self):
        return len(self.cart) == 0

    def total(self):
        return sum(p["price"] for p in self.cart)

    def show(self):
        for p in self.cart:
            print(p["name"], p["price"])

    def clear(self):
        self.cart.clear()
```

```python
# W8A6
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

    def perimeter(self):
        return 2 * (self.w + self.h)

    def scale(self, k):
        self.w *= k
        self.h *= k


w, h, k = map(float, input().split())
r = Rectangle(w, h)
r.scale(k)
print(f"{r.area():.2f} {r.perimeter():.2f}")
```

```python
# W8A7
import math

class Fraction:
    def __init__(self, num, den):
        if den < 0:
            num, den = -num, -den
        g = math.gcd(num, den)
        self.num = num // g
        self.den = den // g

    def add(self, o):
        return Fraction(self.num*o.den + o.num*self.den, self.den*o.den)

    def sub(self, o):
        return Fraction(self.num*o.den - o.num*self.den, self.den*o.den)

    def mul(self, o):
        return Fraction(self.num*o.num, self.den*o.den)

    def div(self, o):
        return Fraction(self.num*o.den, self.den*o.num)

    def __str__(self):
        return f"{self.num}/{self.den}"


a, b, op, c, d = input().split()
a, b, c, d = int(a), int(b), int(c), int(d)
f1, f2 = Fraction(a, b), Fraction(c, d)

res = {
    "+": f1.add,
    "-": f1.sub,
    "*": f1.mul,
    "/": f1.div
}[op](f2)

print(res)
```

```python
# W8A8
class BankAccount:
    def __init__(self, owner, balance):
        self.balance = balance

    def deposit(self, x):
        self.balance += x

    def withdraw(self, x):
        if x <= self.balance:
            self.balance -= x


owner, bal = input().split()
acc = BankAccount(owner, float(bal))

n = int(input())
for _ in range(n):
    cmd, x = input().split()
    x = float(x)
    if cmd == "DEPOSIT":
        acc.deposit(x)
    else:
        acc.withdraw(x)

print(f"{acc.balance:.2f}")
```

```python
# W8A9
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, s):
        self.scores.append(s)

    def avg(self):
        return sum(self.scores) / len(self.scores)

    def rank(self):
        a = self.avg()
        if a >= 8: return "Excellent"
        if a >= 6.5: return "Good"
        if a >= 5: return "Average"
        return "Poor"


name = input()
n = int(input())
st = Student(name)

for _ in range(n):
    _, score = input().split()
    st.add_score(float(score))

print(f"{st.name} {st.avg():.2f} {st.rank()}")
```

```python
# W8A10
class Book:
    def __init__(self, t, a, y):
        self.title = t
        self.author = a
        self.year = y


class Library:
    def __init__(self):
        self.books = []

    def add(self, b):
        self.books.append(b)

    def count_by_author(self, name):
        return sum(1 for b in self.books if b.author == name)

    def find_by_year(self, y):
        return sum(1 for b in self.books if b.year == y)


lib = Library()
n = int(input())
for _ in range(n):
    line = input()
    if line.startswith("ADD"):
        t, a, y = line[4:].split(";")
        lib.add(Book(t, a, int(y)))
    elif line.startswith("COUNTYEAR"):
        print(lib.find_by_year(int(line.split()[1])))
    else:
        print(lib.count_by_author(line.split()[1]))
```

```python
# W8A11
import math

def vec(a, b):
    return [b[i] - a[i] for i in range(3)]

def cross(u, v):
    return [
        u[1]*v[2] - u[2]*v[1],
        u[2]*v[0] - u[0]*v[2],
        u[0]*v[1] - u[1]*v[0]
    ]

def dot(u, v):
    return sum(u[i]*v[i] for i in range(3))

def norm(v):
    return math.sqrt(dot(v, v))


A = list(map(float, input().split()))
B = list(map(float, input().split()))
C = list(map(float, input().split()))
D = list(map(float, input().split()))

n1 = cross(vec(A, B), vec(A, C))
n2 = cross(vec(B, C), vec(B, D))

phi = math.degrees(math.acos(dot(n1, n2) / (norm(n1)*norm(n2))))
print(f"{phi:.2f}")
```

```python
# W8A12
class Complex:
    def __init__(self, r, i):
        self.r = r
        self.i = i

    def multiply(self, o):
        return Complex(self.r*o.r - self.i*o.i,
                       self.r*o.i + self.i*o.r)


a, b = map(float, input().split())
c, d = map(float, input().split())
z1, z2 = Complex(a, b), Complex(c, d)
res = z1.multiply(z2)
print(res.r, res.i)
```

```python
# W8A13
import math

class Triangle:
    def __init__(self, a, b, c):
        self.a, self.b, self.c = a, b, c

    def getArea(self):
        if self.a + self.b <= self.c or self.a + self.c <= self.b or self.b + self.c <= self.a:
            return -1
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p*(p-self.a)*(p-self.b)*(p-self.c))


a, b, c = map(int, input().split())
t = Triangle(a, b, c)
area = t.getArea()

if area == -1:
    print("invalid")
else:
    print(f"{area:.2f}")
```
