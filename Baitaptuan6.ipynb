```python
#W6A1
arr = list(map(int, input().split()))
res = []
seen = set()
for x in arr:
    if x not in seen:
        seen.add(x)
        res.append(x)
print(res)
```

```python
#W6A5
arr = list(map(int, input().split()))
pos = neg = zero = 0
for x in arr:
    if x > 0:
        pos += 1
    elif x < 0:
        neg += 1
    else:
        zero += 1
print({'positives': pos, 'negatives': neg, 'zeros': zero})
```

```python
#W6A9
d1 = eval(input())
d2 = eval(input())
res = {}
for k in set(d1) | set(d2):
    res[k] = d1.get(k, 0) + d2.get(k, 0)
print(dict(sorted(res.items())))
```

```python
#W6A10
arr = list(map(int, input().split()))
k = int(input())
res = []
n = len(arr)
for i in range(n):
    for j in range(i+1, n):
        if arr[i] + arr[j] == k:
            res.append((arr[i], arr[j]))
res.sort()
print(res)
```

```python
#W6A12
arr = list(map(int, input().split()))
freq = {}
for x in arr:
    freq[x] = freq.get(x, 0) + 1
mx = max(freq.values())
candidates = [k for k, v in freq.items() if v == mx]
print(min(candidates))
```

```python
#W6A15
items = input().split()
k = int(input())
d = {}
for i in range(0, len(items), 2):
    key = items[i]
    val = int(items[i+1])
    if val > k:
        d[key] = val
print(d)
```

```python
#W6A18
n, m, k = map(int, input().split())
a = []
for _ in range(n):
    a.append(list(map(int, input().split())))
col_sum = 0
row_sum = 0
for i in range(n):
    col_sum += a[i][k]
for j in range(m):
    row_sum += a[k][j]
print(col_sum)
print(row_sum)
```

```python
#W6A19
def index_of_max(lst):
    mx = lst[0]
    idx = 0
    for i in range(1, len(lst)):
        if lst[i] > mx:
            mx = lst[i]
            idx = i
    return idx
```

```python
#W6A20
def find_first_position(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1
```
