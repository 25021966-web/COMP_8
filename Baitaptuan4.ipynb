```python
#W4A8
def rev(x):
    r = 0
    while x > 0:
        r = r*10 + x%10
        x //= 10
    return r

n = int(input())
step = 0
while True:
    if n == rev(n):
        break
    n += rev(n)
    step += 1
print(step, n)
```

```python
#W4A9
def collatz_len(x):
    cnt = 1
    while x != 1:
        if x % 2 == 0:
            x //= 2
        else:
            x = 3*x + 1
        cnt += 1
    return cnt

n = int(input())
best_x = 1
best_len = 1
for i in range(1, n+1):
    l = collatz_len(i)
    if l > best_len:
        best_len = l
        best_x = i
print(best_x, best_len)
```

```python
#W4A13
def sum_div(x):
    s = 0
    for i in range(1, x):
        if x % i == 0:
            s += i
    return s

a, b = map(int, input().split())
print("true" if sum_div(a) == b and sum_div(b) == a else "false")

```
